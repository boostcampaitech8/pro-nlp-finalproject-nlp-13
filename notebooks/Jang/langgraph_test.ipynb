```python
### ChatPromptTemplate
# {변수} 가 포함된 템플릿을 만듬.
# 자동으로 SystemMessage, HumanMessage 객체 리스트를 반환해줌.

# 만약 ROUTE라면...?
SYSTEM_PROMPT_ROUTE = """
오늘 날짜는 다음과 같습니다: {{current_time}}
일주일은 다음 요일을 순서대로 포함합니다.: {{days}}
당신은 사용자 질문을 [rag, web, weather, direct]로 분류하는 Router입니다.
아래 규칙을 엄격히 준수하여 JSON을 생성하세요.

**★핵심 제약 사항 (위반 시 오답 처리):**
1. **개수 제한:** 각 카테고리(주제) 당 **가장 정확한 '단 하나(1개)'의 쿼리**만 생성하세요.
2. **중복 금지:** 같은 의미의 질문을 여러 번 쓰지 마세요. (유의어 나열 금지)
3. **날씨 포맷:** weather_queries는 반드시 숫자 형식이어야 합니다.

**분류 규칙:**
1. **rag:** 장소, 숙박, 맛집, 명소 정보 -> 리스트에 **핵심 키워드 1개**만.
2. **web:** 실시간 정보(교통편 시간표, 티켓 예매, 뉴스) -> 리스트에 **핵심 문장 1개**만.
3. **weather:** 날씨/기온 -> (내일=1, 오늘=0, 이틀후=2) 형식.
4. **direct:** 인사, 농담.

**출력 예시 (반드시 이 형태를 따를 것):**
User: "내일 부산 날씨랑 해운대 깨끗한 호텔 추천해주고 서울 가는 기차표 제일 빠른거 알려줘"
Output:
{{
  "rag_queries": ["해운대 깨끗한 호텔"], 
  "web_queries": ["서울행 부산 출발 기차표 최단시간"],
  "weather_queries": [1],
  "direct": null,
  "reason": 짧은 설명
}}
"""
```

```python
### 1. 템플릿 정의
prompt_route = ChatPromptTemplate.from_messages([
    ("system", SYSTEM_PROMPT_ROUTE),
    ("human", "{user_input}")
])

print(prompt_route)
print(prompt_route.messages)
```

```python
### 2. 연결
chain = prompt_route | vllm_model

print(chain)
```

### 최종 정리
1. Messages
- 메세지 객채
    - SystemMessage: 감독/설정 | 모델의 페르소나, 규칙, 지식 등
    - HumanMessage: 사용자 | 실제 사용자의 질문이나 입력값
    - AIMessage: AI 모델 | 모델의 이전 답변

1. ChatPromptTemplate
- messages = [SystemMessage(content="..."), HumanMessage(content="...")] -> 이걸 변수로 관리해줌
- ChatPromptTemplate.from_messages([("system", "...")])

```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage, SystemMessage,AIMessage
from langchain_core.prompts import ChatPromptTemplate

### Model
vllm_model = init_chat_model(
    model="dpo-final-policy",  
    model_provider="openai",  
    api_key="none",                    
    #base_url="http://localhost:8000/v1",
    base_url="http://127.0.0.1:8080/v1",
    temperature=0.2,
    max_tokens=2048
)

### 프롬프트
# 이건 .yaml 이나 .txt로 관리
SYSTEM_PROMPT_ROUTE = """
오늘 날짜는 다음과 같습니다: {current_time}
일주일은 다음 요일을 순서대로 포함합니다.: {days}
당신은 사용자 질문을 [rag, web, weather, direct]로 분류하는 Router입니다.
아래 규칙을 엄격히 준수하여 JSON을 생성하세요.

**★핵심 제약 사항 (위반 시 오답 처리):**
1. **개수 제한:** 각 카테고리(주제) 당 **가장 정확한 '단 하나(1개)'의 쿼리**만 생성하세요.
2. **중복 금지:** 같은 의미의 질문을 여러 번 쓰지 마세요. (유의어 나열 금지)
3. **날씨 포맷:** weather_queries는 반드시 숫자 형식이어야 합니다.

**분류 규칙:**
1. **rag:** 장소, 숙박, 맛집, 명소 정보 -> 리스트에 **핵심 키워드 1개**만.
2. **web:** 실시간 정보(교통편 시간표, 티켓 예매, 뉴스) -> 리스트에 **핵심 문장 1개**만.
3. **weather:** 날씨/기온 -> (내일=1, 오늘=0, 이틀후=2) 형식.
4. **direct:** 인사, 농담.

**출력 예시 (반드시 이 형태를 따를 것):**
User: "내일 부산 날씨랑 해운대 깨끗한 호텔 추천해주고 서울 가는 기차표 제일 빠른거 알려줘"
Output:
{{
  "rag_queries": ["해운대 깨끗한 호텔"], 
  "web_queries": ["서울행 부산 출발 기차표 최단시간"],
  "weather_queries": [1],
  "direct": null,
  "reason": 짧은 설명
}}
"""

# Prompt 템플릿
prompt_route = ChatPromptTemplate.from_messages([
    ("system", SYSTEM_PROMPT_ROUTE),
    ("human", "{user_input}")
])

print(prompt_route)

# 템플릿 생성 후 미리 일부만 채움
prompt_route = prompt_route.partial(
    current_time="2026-02-06",
    days = "월, 화, 수, 목, 금, 토, 일",
)

print(prompt_route)

print("[TEST]=============================================")

## 방법 1
chain = prompt_route | vllm_model
response1 = chain.invoke({"user_input": "내일 부산 날씨랑 맛집 알려줘"})
print(response1.content)
print("====================================================")

## 방법 2
messages = prompt_route.format_messages(user_input="내일 부산 날씨랑 맛집 알려줘")
print(messages)
response2 = vllm_model.invoke(messages)
print(response2.content)
print("====================================================")

## 방법 3
chain = prompt_route | vllm_model
print(chain)
for chunk in chain.stream({"user_input": "내일 부산 날씨랑 맛집 알려줘"}):
    print(chunk.content, end="", flush=True)
```
