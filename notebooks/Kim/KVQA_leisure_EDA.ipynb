```python
import glob
import pandas as pd
import os

file_list = glob.glob('./data/220-1.관광 KVQA 데이터(동부권)/01-1.정식개방데이터/Training/01.원천데이터/TS_트리플_동부권_부산_레포츠/*.csv')

rows = []

for file_path in file_list:
    try:
        temp_df = pd.read_csv(file_path, encoding='utf-8')
        row_dict = temp_df.set_index('관광타입')['레포츠'].to_dict()
        rows.append(row_dict)
    except Exception as e:
        print(f"Error parsing {file_path}: {e}")

df = pd.DataFrame(rows)

cols = ['시설명'] + [c for c in df.columns if c not in ['시설명']]
df = df[cols]

print(f"총 {len(df)}개의 데이터가 정리되었습니다.")
df.head()
```

```python
# 부산
df = df[df['주소'].str.contains('부산', na=False)]

print(f"{len(df)}")
df.head()
```

```python
# 리뷰 없는 애들 빼기
df = df[df['리뷰키워드'] != '없음'].copy()

df = df[df['리뷰키워드'] != '없음']
df = df[df['리뷰키워드'].str.len() > 5].copy()

print(f"after filtering: {len(df)}")
```

```python
df.sample(5)
```

```python
# 리뷰 길이
df_sorted = df.sort_values(by='리뷰키워드', key=lambda x: x.str.len(), ascending=True)
df_sorted.head(10)
```

```python
# 헬스, pt, 요가, 필라테스 빼기 
keywords = ['헬스', 'pt', '필라테스', '골프', '볼링', '요가', 'CC', '휘트니스', '짐', '피티', '피트니스', '크로스핏', '테니스', 'GYM']
pattern = '|'.join(keywords) 

# 2. 키워드가 포함되지 않은(~) 행만 뽑기
# case=False : 대소문자 무시 (PT, pt, Pt 다 걸러냄)
# na=False   : 혹시 내용이 비어있어도 에러 안 나게 처리
df_filtered = df[~df['시설명'].str.contains(pattern, case=False, na=False)]

df_filtered = df[~df['체험프로그램'].str.contains(pattern, case=False, na=False)]
df_filtered = df[~df['개요'].str.contains(pattern, case=False, na=False)]

df_filtered = df_filtered[~df_filtered['시설명'].str.contains(pattern, case=False, na=False)]

# 결과 확인
print(f"원래 {len(df)}개 -> 필터링 후 {len(df_filtered)}개")
df_filtered.sample(5)
```
