```python
import glob
import pandas as pd
import os

file_list = glob.glob('./data/220-1.관광 KVQA 데이터(동부권)/01-1.정식개방데이터/Training/01.원천데이터/TS_트리플_동부권_부산_숙박/*.csv')

rows = []

for file_path in file_list:
    try:
        temp_df = pd.read_csv(file_path, encoding='utf-8')
        row_dict = temp_df.set_index('관광타입')['숙박'].to_dict()
        rows.append(row_dict)
    except Exception as e:
        print(f"Error parsing {file_path}: {e}")

df = pd.DataFrame(rows)

cols = ['업소명'] + [c for c in df.columns if c not in ['업소명']]
df = df[cols]

print(f"총 {len(df)}개의 데이터가 정리되었습니다.")
df.head()
```

```python
# 부산시만
df = df[df['주소'].str.contains('부산', na=False)]

print(f"{len(df)}")
df.head()

# 다 부산
```

```python
df = df[df['리뷰키워드'] != '없음'].copy()

df = df[df['리뷰키워드'] != '없음']
df = df[df['리뷰키워드'].str.len() > 5].copy()

print(f"after filtering: {len(df)}")
```
