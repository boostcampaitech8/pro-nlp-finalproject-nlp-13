```python
import pandas as pd
import os
```

```python
data_path = "../../data/persona_data/empathy_multiturn_qa_dataset.csv"
df = pd.read_csv(data_path, usecols=['situation', 'context', 'response'])
```

```python
df
```

```python
for col in ['situation', 'context', 'response']:
    df[col] = df[col].str.replace('\n', ' ', regex=False)
    df[col] = df[col].str.replace('\t', ' ', regex=False)
```

```python
df.to_csv(data_path)
```
